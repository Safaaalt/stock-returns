```python
import pandas as pd
import matplotlib.pyplot as plt
import yfinance as yf
data = yf.download("AAPL", start="2023-01-01",end="2025-01-01",auto_adjust=True)
prices = data["Close"].copy()
print(prices.head())
```

```python
df = prices.copy()
df["daily_return"] = df.iloc[:,0].pct_change()
df = df.dropna()
df.head()
```

```python
print('Number of days :', len(df))
print('Average daily return:',df['daily_return'].mean())
print('Risk(volatility):',df['daily_return'].std())
print('lowest price:',df.iloc[:,0].min())
print('Highest price:',df.iloc[:,0].max())
```

```python
df["daily_return"].plot(title="Daily return percentage",color="orange")
plt.xlabel("Date")
plt.ylabel("Return")
plt.grid(True)
plt.show()
```
